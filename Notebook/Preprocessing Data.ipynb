### processing one-hot encoding for categorical values  
Imputed for missing categorical value  
and Binned using Rare label technique  
Created new features with one hot encoding  Resulting in 199 features.

```python
col = ['NAME_INCOME_TYPE',
    'OCCUPATION_TYPE',
    'HOUSETYPE_MODE',
    'WALLSMATERIAL_MODE',
    'NAME_CONTRACT_TYPE'
]
rare_threshold = 0.02
for i in col:
    if data[i].isnull().sum() > 0:
        data[i] = data[i].fillna('Unknown')
    freqs = data[i].value_counts(normalize=True)
    rare_labels = freqs[freqs < rare_threshold].index
    if len(rare_labels) > 0:
        data[i] = data[i].replace(rare_labels, 'Rare')
data_encoded = pd.get_dummies(data, columns=col, drop_first=True)
```

### Checking Lable Names and removing any special character on feature names

```python
# def clean_feature_name(name):
colume = []
for name in data_encoded.columns:
    """Remove special JSON characters from feature names"""
    # Replace special characters with underscore
    name = re.sub(r'[^\w\s]', '_', str(name))
    # Remove multiple consecutive underscores
    name = re.sub(r'_+', '_', name)
    # Remove leading/trailing underscores
    name = name.strip('_')
    colume.append(name)
data_encoded.columns = colume
    

# # Clean column names
# X_train.columns = [clean_feature_name(col) for col in X_train.columns]
# X_test.columns = [clean_feature_name(col) for col in X_test.columns]
# X_train_scaled.columns = [clean_feature_name(col) for col in X_train_scaled.columns]
# X_test_scaled.columns = [clean_feature_name(col) for col in X_test_scaled.columns]

# print("✅ Feature names cleaned!")
# print(f"Sample cleaned names: {list(X_train.columns[:5])}")
```
